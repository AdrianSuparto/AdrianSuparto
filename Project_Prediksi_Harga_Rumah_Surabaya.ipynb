###2.1 Data Wrangling dan Transformasi

Data wrangling atau dikenal juga sebagai data munging adalah proses mempersiapkan dan membersihkan data mentah agar dapat digunakan dalam analisis atau pemodelan machine learning (Kandel et al., 2011). Proses ini melibatkan penggabungan sumber data, menghapus data duplikat, menangani missing value, hingga mengubah format data menjadi bentuk yang terstruktur dan konsisten.

Dalam proyek prediksi harga rumah, data yang berasal dari scraping web seperti rumah123.com seringkali tidak memiliki struktur yang ideal untuk analisis. Data wrangling menjadi langkah krusial untuk memastikan kualitas data yang akan memengaruhi hasil model.

Setelah data dibersihkan, tahap berikutnya adalah transformasi data, yaitu proses mengubah data ke dalam format atau struktur yang sesuai untuk analisis lebih lanjut (Kimball & Ross, 2013). Transformasi meliputi:

- Normalisasi atau standardisasi nilai numerik (misalnya luas bangunan dalam m²)

- Encoding fitur kategorikal (misal: jenis sertifikat atau tipe rumah)

- Feature engineering (misalnya membuat fitur baru seperti harga_per_meter atau usia_bangunan)

Transformasi penting karena:

- Model statistik atau machine learning memiliki asumsi tertentu terhadap data.

- Fitur baru dapat meningkatkan kemampuan model dalam menangkap pola.

Beberapa pustaka dan alat umum untuk wrangling dan transforming:

- Python + Pandas: Library utama untuk manipulasi data tabular (McKinney, 2012).

- NumPy: Untuk transformasi numerik.

- Scikit-learn: Menyediakan pipeline transformasi seperti StandardScaler, OneHotEncoder, dll.

- OpenRefine atau Excel: Untuk eksplorasi awal data dalam bentuk tabel.

Contoh transformasi pada proyek prediksi harga rumah:

- Mengonversi kolom “harga” dari string “Rp 1,2 Miliar” menjadi angka 1200000000.

- Mengelompokkan kolom “sertifikat” menjadi kategori ordinal.

- Membuat kolom harga_per_m2 sebagai harga / luas_bangunan.

Menurut Han et al. (2011), kualitas model sangat bergantung pada kualitas input data. Jika data tidak ditangani dengan benar pada tahap wrangling dan transformasi, bahkan algoritma machine learning terbaik pun akan gagal memberikan hasil yang baik.

Kesalahan umum dalam tahapan ini meliputi:

- Tidak menangani outlier

- Salah melakukan encoding kategorikal

- Menghilangkan baris yang sebenarnya penting saat cleaning

- Membiarkan missing value tanpa perlakuan khusus

Data wrangling dan transformasi bukan hanya bagian awal yang "membosankan", tetapi justru pondasi dari keseluruhan pipeline data science. Dalam konteks prediksi harga rumah, proses ini menentukan sejauh mana data mentah bisa digunakan untuk mengekstrak wawasan dan membangun model prediktif yang akurat.

Tanpa wrangling yang rapi dan transformasi yang tepat, model akan memiliki bias, performa buruk, dan berpotensi menghasilkan insight yang menyesatkan.

###2.5 ANOVA (Analysis of Variance)

ANOVA (Analysis of Variance) adalah teknik statistik inferensial yang digunakan untuk menguji apakah terdapat perbedaan yang signifikan secara statistik antara rata-rata dua atau lebih kelompok independen. Metode ini dikembangkan oleh Ronald A. Fisher pada awal abad ke-20 dan telah menjadi salah satu metode utama dalam analisis perbedaan kelompok.

Tujuan utama ANOVA adalah untuk mengetahui apakah variasi dalam variabel dependen (misalnya: harga rumah) dapat dijelaskan oleh variasi dalam variabel independen kategorikal (misalnya: jenis sertifikat, tipe rumah, atau kecamatan).

ANOVA membagi total variasi dalam data menjadi dua komponen:

- Antar-kelompok (Between-group variance): Variasi rata-rata antar kelompok.

- Dalam-kelompok (Within-group variance): Variasi data di dalam kelompok yang sama.

Statistik uji yang digunakan adalah F-ratio, yaitu perbandingan antara varians antar-kelompok dengan varians dalam-kelompok:

image.png

Jika nilai F cukup besar dan signifikan secara statistik (p-value < 0.05), maka dapat disimpulkan bahwa terdapat perbedaan rata-rata yang signifikan antar kelompok.

####4.1.2 ANOVA

```python
from scipy.stats import f_oneway

def anova_feature_selection(df: pd.DataFrame, categorical_column: str, target_column: str, alpha: float = 0.05):
    """
    Menggunakan ANOVA satu arah untuk menguji apakah fitur kategorikal memiliki pengaruh signifikan terhadap target numerik.

    Parameters:
    - df (pd.DataFrame): DataFrame input.
    - categorical_column (str): Nama kolom kategorikal (nominal).
    - target_column (str): Nama kolom target numerik.
    - alpha (float): Signifikansi (default 0.05).

    Returns:
    - p_value (float): Nilai p hasil uji ANOVA.
    - significant (bool): Apakah hasil signifikan pada level alpha.
    - groups (dict): Dictionary kategori dengan nilai target masing-masing.
    """
    # Kelompokkan target berdasarkan kategori
    groups = {
        cat: df[df[categorical_column] == cat][target_column].dropna().values
        for cat in df[categorical_column].dropna().unique()
    }

    # Pastikan ada minimal dua grup untuk diuji
    if len(groups) < 2:
        raise ValueError("Minimal diperlukan dua kategori untuk uji ANOVA.")

    # Uji ANOVA
    f_stat, p_value = f_oneway(*groups.values())

    return {
        "categorical_column": categorical_column,
        "p_value": p_value,
        "significant": p_value < alpha,
        "alpha": alpha,
        "group_summary": {k: len(v) for k, v in groups.items()}
    }
```

####4.4.2 Local Modeling

```python
# Simpan hasil model dan skor
model_per_kecamatan = {}
skor_per_kecamatan = {}

# Loop setiap kecamatan
for kecamatan in df['Kecamatan'].unique():
    df_kec = df[df['Kecamatan'] == kecamatan]

    # Pastikan semua fitur tersedia di kecamatan ini
    fitur_tersedia = [f for f in df_fitur if f in df_kec.columns]

    if len(fitur_tersedia) == 0:
        print(f"Kecamatan {kecamatan} tidak memiliki fitur global, lewati.")
        continue

    x = pd.get_dummies(df_kec[fitur_tersedia], drop_first=True)
    y = df_kec['log_price']

    if len(x) < 10:
        print(f"Data terlalu sedikit di {kecamatan}, lewati.")
        continue

    # Split & training
    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
    model_local = LinearRegression()
    model_local.fit(x_train, y_train)

    # Simpan model
    model_per_kecamatan[kecamatan] = model_local

    # Prediksi dan konversi kembali dari log ke harga asli
    y_pred = model_local.predict(x_test)
    y_pred_exp = np.exp(y_pred)
    y_test_exp = np.exp(y_test)

    # Evaluasi di skala harga asli
    skor_per_kecamatan[kecamatan] = {
        'R2': r2_score(y_test_exp, y_pred_exp),
        'MAE': mean_absolute_error(y_test_exp, y_pred_exp),
        'MSE': mean_squared_error(y_test_exp, y_pred_exp),
        'RMSE': np.sqrt(mean_squared_error(y_test_exp, y_pred_exp)),
        'MAPE': np.mean(np.abs((y_test_exp - y_pred_exp) / y_test_exp)) * 100
    }

# Tampilkan skor semua kecamatan
print(pd.DataFrame(skor_per_kecamatan).T)
```

✅ Model dengan performa bagus:
1. semampir:

 - R²: 0.92, sangat tinggi (model menjelaskan 92% variasi data)
 - MAPE: 6.07%, sangat kecil (akurasi sangat baik)

2. sukolilo, dukuh pakis, rungkut, tenggilis mejoyo:

 - R² antara 0.71–0.84  
 - MAPE di bawah 10%, akurasi baik.

⚠️ Model performa sedang:
1. kenjeran, wonocolo, pakal, mulyorejo, tambaksari:

- R² antara 0.4–0.6, MAPE sekitar 9–11%, cukup layak tetapi bisa lebih baik.

❌ Model performa buruk:
1. wonokromo, wiyung, sukomanunggal, gayungan, gubeng:

 - R² sangat rendah atau negatif → model tidak mampu menjelaskan data

 - MAPE tinggi (>13%), prediksi tidak akurat.

 - Misalnya gayungan → R² = -0.11 → model lebih buruk dari rata-rata!

```python
residual_stats = []

for kecamatan in model_per_kecamatan.keys():
    model = model_per_kecamatan[kecamatan]
    df_kec = df[df['Kecamatan'] == kecamatan]

    fitur_tersedia = [f for f in df_fitur if f in df_kec.columns]
    if len(fitur_tersedia) == 0 or len(df_kec) < 10:
        continue

    x = pd.get_dummies(df_kec[fitur_tersedia], drop_first=True)
    y = df_kec['log_price']

    x_train, x_test, y_train, y_test = train_test_split(
        x, y, test_size=0.2, random_state=42
    )
    y_test_pred = model.predict(x_test)

    # Kembalikan dari log ke harga
    y_test_actual = np.exp(y_test)
    y_test_pred_actual = np.exp(y_test_pred)

    # Hitung residual
    residual_pct = 100 * (y_test_actual - y_test_pred_actual) / y_test_pred_actual
    mean_resid = residual_pct.mean()
    std_resid = residual_pct.std()

    # Simpan hasil
    residual_stats.append({
        'Kecamatan': kecamatan,
        'Mean Residual (%)': round(mean_resid, 2),
        'Std Residual (%)': round(std_resid, 2)
    })

# Cetak hasil
residual_df = pd.DataFrame(residual_stats).set_index('Kecamatan')
print(residual_df)
```

📊 Interpretasi Kolom
Mean Residual (%): Rata-rata selisih prediksi terhadap harga sebenarnya dalam persen.

- Positif → prediksi lebih rendah dari harga sebenarnya (underestimasi).

- Negatif → prediksi lebih tinggi dari harga sebenarnya (overestimasi).

- Std Residual (%): Seberapa besar sebaran (variabilitas) error prediksi — semakin besar artinya model kurang stabil.

| Kecamatan            | Mean Residual | Std Dev | Interpretasi Singkat                        |
| -------------------- | ------------- | ------- | ------------------------------------------- |
| **semampir**         | -1.63%        | 8.28%   | Akurat, prediksi sedikit lebih tinggi       |
| **wonokromo**        | +0.51%        | 29.73%  | Variasi prediksi besar, model tidak stabil  |
| **tambaksari**       | +1.16%        | 16.44%  | Cukup stabil, sedikit underestimasi         |
